# Email Filtering using Set Theory

```python
# Email sets
all_emails = {"email1", "email2", "email3", "email4", "email5", "email6"}
spam_emails = {"email2", "email4"}
important_emails = {"email1", "email5"}
duplicate_emails = {"email3", "email3"}  # Simulated via repetition
read_emails = {"email1", "email2"}
```

# Set Operations

Union (A ∪ B) – Everything in A or B or both. Basically, the more the merrier.

```python
all_notifications = important_emails.union(spam_emails)
```

# Intersection (A ∩ B) – Only the stuff in both A and B. Where the magic overlaps.

```python
important_and_read = important_emails.intersection(read_emails)
```

# Difference (A − B or B − A) – What’s in A but not in B (or vice versa). Like unfriending someone.

```python
filtered_emails = all_emails.difference(spam_emails)
```

# Complement (A′ or ~A) – Everything not in A (assuming a universal set U). A's arch-nemesis.

```python
unread_emails = all_emails.difference(read_emails)
```

# Symmetric Difference (A △ B) – In A or B, but not in both. Mutual exes.

```python
attention_required = important_emails.symmetric_difference(spam_emails)
```

# Display Output of All operations

```python
print("All Notifications:", all_notifications)
print("Important and Read:", important_and_read)
print("Filtered Emails (No Spam):", filtered_emails)
print("Unread Emails:", unread_emails)
print("Attention Required Emails:", attention_required)
```
